## 1. Resumen del proyecto

El objetivo del proyecto fue construir una pipeline Medallón para analizar patrones de fraude financiero usando PySpark y Databricks.

La solución se organizó en tres capas:

- **Bronze:** ingesta de las fuentes originales como tablas Delta.
- **Silver:** limpieza, estandarización, JOINs y enriquecimiento de transacciones.
- **Gold:** tablas analíticas orientadas a negocio para análisis de fraude y perfilamiento de usuarios.

Las tablas principales creadas fueron:

- `workspace.bronze.transactions_daniel`
- `workspace.bronze.users_daniel`
- `workspace.bronze.cards_daniel`
- `workspace.bronze.mcc_codes_daniel`
- `workspace.bronze.fraud_labels_daniel`
- `workspace.silver.transactions_daniel`
- `workspace.gold.resumen_fraude_daniel`
- `workspace.gold.fraude_por_dimension_daniel`
- `workspace.gold.fraude_por_monto_daniel`
- `workspace.gold.perfil_usuario_daniel`

## 2. Tasa de fraude global

El dataset Silver contiene **13,305,915 transacciones**.

De estas:

- **8,914,963** tienen etiqueta de fraude.
- **4,390,952** no tienen label.
- **13,332** fueron fraudulentas.
- **8,901,631** fueron legítimas.

La tasa global de fraude calculada sobre transacciones etiquetadas fue:

**0.1495%**

En términos de monto:

- Monto total fraudulento: **1,749,004.78**
- Monto total transaccionado: **706,873,606.96**
- Porcentaje del monto fraudulento: **0.2474%**

Aunque la proporción de fraudes es baja en cantidad de transacciones, el porcentaje del monto fraudulento es ligeramente mayor, lo que indica que el impacto económico proporcional del fraude es más alto que su frecuencia.

## 9. Decisiones tomadas en Silver que afectaron Gold

Las principales decisiones tomadas en Silver fueron:

1. **Uso de LEFT JOINs:**  
   Se conservaron todas las transacciones originales, incluso si alguna dimensión no tenía coincidencia.

2. **Conversión de `target` a `is_fraud`:**  
   Se convirtió el label `Yes/No` a valores numéricos `1/0` para permitir agregaciones.

3. **Conservación de registros sin label:**  
   Las transacciones sin etiqueta no se eliminaron. Se conservaron en Silver, pero en Gold las tasas se calcularon usando solo transacciones etiquetadas.

4. **Creación de `amount_abs`:**  
   Se usó para analizar volumen financiero sin eliminar montos negativos, que pueden representar reversas o devoluciones.

5. **Conversión de MCC a formato largo:**  
   Esto permitió unir correctamente las transacciones con categorías de comercio y construir análisis por `merchant_category`.

Estas decisiones fueron importantes porque afectaron directamente las métricas de fraude, las tasas calculadas y la interpretación de los resultados.

## 10. Información del dataset que no se pudo usar efectivamente

La principal limitación fue que no todas las transacciones tenían etiqueta de fraude. Había **4,390,952 transacciones sin label**.

Por esta razón, las tasas de fraude se calcularon únicamente sobre las transacciones etiquetadas. Incluir las transacciones sin label en el denominador habría diluido artificialmente la tasa.

También algunas columnas del dataset podrían requerir análisis adicional antes de ser usadas en modelos de riesgo, por ejemplo campos de tarjeta sensibles o datos que no necesariamente aportan valor directo al análisis agregado.
