**Autor:** Daniel Guzmán  
**Actividad:** Actividad 04 — Arquitectura Medallón  
**Capa:** Bronze  
**Entorno:** Databricks

```python
catalog = "workspace"
schema = "default"
volume = "customers_activity_04_daniel"

path_volume = f"/Volumes/{catalog}/{schema}/{volume}"
file_path = f"{path_volume}/semana_01_actividad_04/customers-2000000.csv"

print(file_path)
```

```python
df_bronze = (
    spark.read.format("csv")
    .option("header", "true")
    .option("inferSchema", "true")
    .load(file_path)
)

print(f"Registros: {df_bronze.count()}")
df_bronze.printSchema()
df_bronze.show(5)
```

```python
spark.sql("USE CATALOG workspace")
spark.sql("USE SCHEMA default")

spark.sql("DROP TABLE IF EXISTS bronze_customers")

df_bronze.write.format("delta") \
    .mode("overwrite") \
    .option("delta.columnMapping.mode", "name") \
    .option("delta.minReaderVersion", "2") \
    .option("delta.minWriterVersion", "5") \
    .saveAsTable("bronze_customers")
```

```python
df_check = spark.read.table("workspace.default.bronze_customers")

print(f"Registros en bronze_customers: {df_check.count()}")
df_check.printSchema()
display(df_check.limit(5))
```
